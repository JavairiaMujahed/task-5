Now let's try something special. If we see the name column, there is data which corresponds to the age of the person. Yes: Mr. Mrs. Master. Miss. So lets use that in filling the NA values for age

```python
df.head(5)
```

```python
def extract(x):
    temp = x.split(" ")
    if "Mr." in temp:
        return "Mr"
    elif "Mrs." in temp:
        return "Mrs"
    elif "Miss." in temp:
        return "Miss"
    elif "Master." in temp:
        return "Master"
    elif "Dr." in temp:
        return "Dr"
    else:
        return None
```

```python
df["Category"] = df["Name"].apply(extract)
```

```python
df.head()
```

```python
df["Category"].unique()
```

```python
print("Mr." , np.mean(df[df["Category"] == "Mr"]["Age"]))
print("Mrs." , np.mean(df[df["Category"] == "Mrs"]["Age"]))
print("Miss." , np.mean(df[df["Category"] == "Miss"]["Age"]))
print("Master." , np.mean(df[df["Category"] == "Master"]["Age"]))
print("Dr." , np.mean(df[df["Category"] == "Dr"]["Age"]))
```

```python
male_dead=((df['Sex']=='male')&(df['Survived']==0)).sum(0)
print("Male Dead: ",male_dead)
male_survived=((df['Sex']=='male')&(df['Survived']==1)).sum(0)
print("Male Survived: ",male_survived)

female_dead=((df['Sex']=='female')&(df['Survived']==0)).sum(0)
print("Female Dead: ",female_dead)
female_survived=((df['Sex']=='female')&(df['Survived']==1)).sum(0)
print("Female Survived: ",female_survived)
```

```python
#now we will be plotting a bar graph depicting the values we calculated above
m_data=(male_dead,male_survived)
f_data=(female_dead,female_survived)
index=['Dead','Survived']
p1=plt.bar(np.arange(len(m_data)),m_data,color='blue')
p2=plt.bar(np.arange(len(f_data)),f_data,bottom=m_data,color='orange')
plt.xticks(np.arange(len(m_data)),index)
plt.legend((p1[0],p2[0]),('Male','Female'))
plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Gender Wise Survival')
plt.show()
```
